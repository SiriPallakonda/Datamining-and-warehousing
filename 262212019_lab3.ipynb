```python
import matplotlib.pyplot as plt
import seaborn as sns

x = df["bmi"]

df["z_score"] = (x - x.mean()) / x.std()
df["min_max"] = (x - x.min()) / (x.max() - x.min())
df["mean_norm"] = (x - x.mean()) / (x.max() - x.min())
df["max_absolute"] = x / x.abs().max()

q1 = x.quantile(0.25)
q3 = x.quantile(0.75)
iqr = q3 - q1

df["robust"] = (x - x.median()) / iqr if iqr != 0 else 0
cols = ["z_score", "min_max", "mean_norm",
        "max_absolute", "robust"]

plt.figure(figsize=(8, 6))
sns.heatmap(df[cols].corr(), annot=True,
            cmap="coolwarm", fmt=".2f")

plt.title("Heatmap of Scaling Techniques")
plt.show()
```
